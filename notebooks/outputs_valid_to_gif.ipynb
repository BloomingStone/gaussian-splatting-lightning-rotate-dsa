```python
from pathlib import Path
import imageio.v3 as iio
import numpy as np
from typing import List

def make_gif_from_pngs(
    png_list: List[Path],
    output_path: Path,
    fps_gif: float
) -> None:
    """
    将一系列 PNG 图像合成为 GIF。
    
    Args:
        png_list (List[Path]): PNG 图片路径列表，按顺序排列。
        output_path (Path): 输出 GIF 文件路径。
        fps_gif (float): GIF 帧率（每秒帧数）。
    """
    assert len(png_list) > 0, "png_list 不能为空"
    
    # 读取所有 PNG 图片为 numpy 数组
    frames = [iio.imread(png_path) for png_path in png_list]
    
    frames_np = np.stack(frames)
    
    duration_ms = 1000 / fps_gif
    
    iio.imwrite(output_path, frames_np, extension=".gif", duration=duration_ms)
    
    print(f"✅ GIF 已保存至: {output_path.resolve()}")
```

```python
import itertools
import re

# 定义提取 epoch 的正则
epoch_pattern = re.compile(r"epoch=(\d+)")

def extract_epoch(path: Path) -> int:
    """从文件名中提取 epoch 数值，用于排序 key"""
    match = epoch_pattern.search(path.name)
    return int(match.group(1)) if match else -1  # 没匹配到时返回 -1（或 0）

def output_val_to_gif(
    outputs_name: str, 
    view_id: str = "001", 
    max_frames: int = 50,
    fps_gif: int = 10
):
    root_dir = Path.cwd().parent
    val_dir = root_dir / "outputs" / outputs_name / "val"
    png_list = []
    for d in itertools.islice(sorted(val_dir.iterdir(), key=extract_epoch), max_frames):
        image = d / f"{view_id}.png.jpg"
        if image.exists():
            png_list.append(image)

    output_path = root_dir / "outputs" / "temp" / outputs_name / f"{view_id}.gif"
    output_path.parent.mkdir(exist_ok=True, parents=True)
    make_gif_from_pngs(png_list, output_path, fps_gif=fps_gif)
```

```python
output_names = [
    "rotate-xcat__90-frames",
    "rotate-xcat__x-gs__90-frames",
    "rotate-xcat-static",
    "rotate-xcat-static-label"
]

for output_name in output_names:
    output_val_to_gif(output_name)
```
